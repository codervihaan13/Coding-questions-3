```python
#11.	Write a function factorial(num) that returns the factorial of a number.
def factorial(num):
  if num==0 or num==1:
    return 1
  else:
    return num*factorial(num-1)
num1=int(input("Enter number to find its factorial: "))
print(f"The factorial of {num1} is {factorial(num1)}.")
```

```python
#12.	Write a function sum_digits(num) that returns the sum of all digits of a number.
def sum_digits(num):
    if num < 0:
        num = -num
    return sum(int(digit) for digit in str(num))
number = int(input("Enter an integer: "))
result = sum_digits(number)
print(f"The sum of the digits of {number} is: {result}")
```

```python
#13.	Write a function reverse_number(num) that returns the reverse of a number.
def reverse_number(num):
    num_str = str(num)
    if num_str[0] == '-':
        return int('-' + num_str[:0:-1])
    else:
        return int(num_str[::-1])
number = int(input("Enter an integer: "))
result = reverse_number(number)
print(f"The reverse of {number} is: {result}")
```

```python
#14.	Write a function is_palindrome(num) that checks whether a number is a palindrome.
def palindrome(num):
  if len(num)<=1:
    return True
  if num[0]!=num[-1]:
    return False
  return palindrome(num[1:-1])
word=input("Enter number to find if it is a palindrome or not: ")
if palindrome(word):
  print("It's a palindrome.")
else:
  print("It's not a palindrome.")
```

```python
#15.	Write a function is_prime(num) that checks whether a number is prime.
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True
number = int(input("Enter an integer: "))
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

```python
#16.	Write a function find_hcf(a, b) that returns the HCF/GCD of two numbers using the Euclidean Algorithm.
def find_hcf(a, b):
    while b != 0:
        a, b = b, a % b
    return a
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print(f"The HCF/GCD of {num1} and {num2} is: {find_hcf(num1, num2)}")
```

```python
#17.	Write a function is_armstrong(num) that checks whether a number is an Armstrong number.
def is_armstrong(num):
    num_str = str(num)
    power = len(num_str)
    total_sum = sum(int(digit) ** power for digit in num_str)
    return total_sum == num
number = int(input("Enter a positive number: "))
if is_armstrong(number):
    print(f"{number} is an Armstrong number.")
else:
    print(f"{number} is not an Armstrong number.")
```

```python
#18.	Write a function count_vowels(text) that returns the number of vowels in a string.
def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)
user_text = input("Enter a string: ")
print(f"Number of vowels: {count_vowels(user_text)}")
```

```python
#19.	Write a function find_max(numbers) that accepts a list of numbers and returns the largest number without using Python's built-in max() function.
def find_max(numbers):
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest
count = int(input("How many numbers do you want to enter? "))
numbers = []
for i in range(count):
    num = float(input(f"Enter number {i + 1}: "))
    numbers.append(num)
result = find_max(numbers)
print("The largest number is:", result)
```

```python
#20.	Write a recursive function fibonacci(n) that returns the nth Fibonacci number.
def fibonacci(n):
  if n==0:
    return 0
  if n==1:
    return 1
  return fibonacci(n-1)+fibonacci(n-2)
num=int(input("Enter position: "))
print(f"The fibonacci series is: ")
for i in range(num+1):
   print(fibonacci(i),end=" ")
```
